```python
import numpy as np
import matplotlib.pyplot as plt

studentID = ['S1', 'S2', 'S3']
X = np.array([1, 2, 3], dtype=float)
Y = np.array([2, 3, 4], dtype=float)

m = len(X)
w = 0.0
b = 0.0


studentID = ['S1', 'S2', 'S3']
X = np.array([1, 2, 3], dtype=float)
Y = np.array([2, 3, 4], dtype=float)

m = len(X)


w = 0.0
b = 0.0

learning_rate = 0.05
iterations = 2

cost_history = []


for i in range(iterations):


    Y_pred = w * X + b


    error = Y_pred - Y


    cost = (1 / (2 * m)) * np.sum(error ** 2)
    cost_history.append(cost)


    dw = (1 / m) * np.sum(error * X)
    db = (1 / m) * np.sum(error)

    w = w - learning_rate * dw
    b = b - learning_rate * db


print("Final Weight (w):", w)
print("Final Bias (b):", b)
print("Final Cost:", cost_history[-1])

print("\nLinear Regression Equation:")
print(f"Y = {w:.4f}X + {b:.4f}")


Y_pred = w * X + b

print("\nPredictions:")
for sid, x, actual, predicted in zip(studentID, X, Y, Y_pred):
    print(f"{sid}: X={x}, Actual Y={actual}, Predicted Y={predicted:.4f}")



plt.figure(figsize=(12, 5))


plt.subplot(1, 2, 1)
plt.plot(cost_history, color='blue')
plt.xlabel("Iterations")
plt.ylabel("Cost")
plt.title("Cost Decreasing using Gradient Descent")
plt.grid(True)


plt.subplot(1, 2, 2)
plt.scatter(X, Y, color='red', s=80, label='Actual Data')

x_line = np.linspace(0, 4, 100)
y_line = w * x_line + b

plt.plot(x_line, y_line, color='green',
         label='Regression Line')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()


print("Initial weight: ", w)
print("Initial bias: ", b)
```
