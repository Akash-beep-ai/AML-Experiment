```python
A dataset which consist of 6 observation with 2 feature and a target
observation- 1,2,3,4,5,6 rainfall- high,high,high,low,low,low soiltype- loamy,sandy,clay,loamy,sandy,clay yeild-good,good,good,poor,poor,good
use OneR to determine which feature give the best single rule classify
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score

data = {
    'observation': [1, 2, 3, 4, 5, 6],
    'rainfall': ['high', 'high', 'high', 'low', 'low', 'low'],
    'soiltype': ['loamy', 'sandy', 'clay', 'loamy', 'sandy', 'clay'],
    'yield': ['good', 'good', 'good', 'poor', 'poor', 'good']
}
df = pd.DataFrame(data)

X = df[['rainfall', 'soiltype']]
y = df['yield']

def oner_classifier(X, y):
    best_feature = None
    best_accuracy = -1
    all_rules = {}

    for feature in X.columns:

        rules_for_feature = {}
        for value in X[feature].unique():
            subset_y = y[X[feature] == value]
            if not subset_y.empty:

                prediction = subset_y.mode()[0]
                rules_for_feature[value] = prediction

        y_pred = X[feature].map(rules_for_feature)

        valid_indices = ~y_pred.isna()
        if valid_indices.any():
            accuracy = accuracy_score(y[valid_indices], y_pred[valid_indices])
        else:
            accuracy = 0.0

        all_rules[feature] = {'rules': rules_for_feature, 'accuracy': accuracy}

        if accuracy > best_accuracy:
            best_accuracy = accuracy
            best_feature = feature

    return best_feature, best_accuracy, all_rules

best_feature, best_accuracy, rules = oner_classifier(X, y)

print("OneR Rules:")
for feature, rule_data in rules.items():
    print(f"  Feature: {feature}, Accuracy: {rule_data['accuracy']:.2f}")
    for value, pred in rule_data['rules'].items():
        print(f"    If {feature} is {value}, then yield is {pred}")

print(f"\nThe feature that gives the best single rule classification is: '{best_feature}' with an accuracy of {best_accuracy:.2f}")
```
