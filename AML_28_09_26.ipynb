```python
import pandas as pd
import numpy as np
from itertools import combinations
data = {
    "student": [1, 2, 3, 4, 5, 6],
    "x1": [1, 2, 3, 7, 8, 9],
    "x2": [2, 3, 4, 6, 7, 8],
    "x3": [5, 4, 3, 3, 2, 1],
    "y":  [0, 0, 0, 1, 1, 1]
}
df = pd.DataFrame(data)

# Calculate the correlation matrix
corr = df.corr()

rcf = {
    "x1": abs(corr.loc["x1", "y"]),
    "x2": abs(corr.loc["x2", "y"]),
    "x3": abs(corr.loc["x3", "y"])
}
print("\nRCF (Feature-Class Correlation)")
for feature, value in rcf.items():
    print(f"r_{feature}y = {value:.4f}")
rff = {
    ("x1", "x2"): abs(corr.loc["x1", "x2"]),
    ("x2", "x3"): abs(corr.loc["x2", "x3"]),
    ("x1", "x3"): abs(corr.loc["x1", "x3"])
}
print("\nRFF (Feature-Feature Correlation)")
for pair, value in rff.items():
    print(f"r_{pair[0]}{pair[1]} = {value:.4f}")
def cfs_merit(subset):
    k = len(subset)
    avg_rcf = np.mean(
        [
            rcf[f] for f in subset
        ]
    )
    if k == 1:
        avg_rff = 0
    else:
        f1, f2 = subset
        avg_rff = rff[tuple(sorted((f1, f2)))]
    merit = (k * avg_rcf) / np.sqrt(
        k + k * (k - 1) * avg_rff
    )
    return avg_rcf, avg_rff, merit
subsets = [
    ("x1", "x2"),
    ("x1", "x3"),
    ("x2", "x3")
]
results = []
for subset in subsets:
    avg_rcf, avg_rff, merit = cfs_merit(subset)
    results.append(
        [
            ",".join(subset),
            len(subset),
            avg_rcf,
            avg_rff,
            merit
        ]
    )
result = pd.DataFrame(
    results,
    columns=
        [
            "Candidate Subset",
            "k",
            "Average r_cf",
            "Average r_ff",
            "CFS Merit"
        ]

)
result = result.sort_values(
    "CFS Merit",
    ascending=False
)
print("\n" + "=" * 65)
print("CFS COMPARISON TABLE")
print("=" * 65)
print(result.to_string(
    index=False,
    float_format=lambda x: f"{x:.4f}"
))
best = result.iloc[0]
print("\n" + "=" * 65)
print("BEST SUBSET")
print("=" * 65)
print("Best subset :", best["Candidate Subset"])
print("CFS Merit   :", round(best["CFS Merit"], 4))
```
